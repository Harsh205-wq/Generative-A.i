```python
from nltk import WordNetLemmatizer,PorterStemmer,word_tokenize
```

```python
words = [
    "running",
    "studies",
    "better",
    "wolves",
    "flying",
    "easily",
    "happier",
    "mice",
    "geese",
    "cars",
    "playing",
    "ran",
    "feet",
    "cats",
    "talked"
]
```

```python
stemming=PorterStemmer()
lemmtizing=WordNetLemmatizer()
def compare_normalizers(words):
    result = []
    
    for word in words:
        stem = stemming.stem(word)
        lemma = lemmtizing.lemmatize(word)
        
        status = "same" if stem == lemma else "different"
        
        result.append({
            "word": word,
            "stem": stem,
            "lemma": lemma,
            "status": status
        })
    
    return result
```

```python
print(compare_normalizers(words))
```
