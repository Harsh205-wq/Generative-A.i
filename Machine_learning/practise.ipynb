```python
from nltk import WordNetLemmatizer,PorterStemmer,word_tokenize
```

```python
words = [
    "running",
    "studies",
    "better",
    "wolves",
    "flying",
    "easily",
    "happier",
    "mice",
    "geese",
    "cars",
    "playing",
    "ran",
    "feet",
    "cats",
    "talked"
]
```

```python
stemming=PorterStemmer()
lemmtizing=WordNetLemmatizer()
def compare_normalizers(words):
    result = []
    
    for word in words:
        stem = stemming.stem(word)
        lemma = lemmtizing.lemmatize(word)
        
        status = "same" if stem == lemma else "different"
        
        result.append({
            "word": word,
            "stem": stem,
            "lemma": lemma,
            "status": status
        })
    
    return result
```

```python
print(compare_normalizers(words))
```

```python
%pip install pandas
import pandas as pd
messages = pd.read_csv('spam_messages.csv',sep='\t',names=['label','message'])
```

```python
## Data cleaning and preprocessing
import nltk
nltk.download('stopwords')
nltk.download("wordnet", quiet=True)
nltk.download("omw-1.4", quiet=True)
```

```python
from nltk.stem import WordNetLemmatizer
import re
lemmatizer = WordNetLemmatizer()
```

```python
from nltk.corpus import stopwords

corpus=[]
for i in range (0,len(messages)):
     review=re.sub('[^a-zA-Z]',' ',messages['message'][i])
     review.lower()
     review=review.split()
     review=[lemmatizer.lemmatize(word) for word in review if not word in stopwords.words('english')]
     review=' '.join(review)
     corpus.append(review)
    
```

```python
corpus
```

```python
## Create the Bag of Words model
%pip install scikit-learn
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=100, binary=True)
```

```python
X=cv.fit_transform(corpus).toarray()
```
