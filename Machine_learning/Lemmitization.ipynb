Lemmatization


```python
import nltk

nltk.download("wordnet", quiet=True)
nltk.download("omw-1.4", quiet=True)
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

print(lemmatizer.lemmatize("going"))
print(lemmatizer.lemmatize("going", pos="v"))
```

```python
words = [
    "eat", "eating", "eaten", "eats",
    "final", "finally", "finalize", "finalized",
    "play", "playing", "played", "player",
    "study", "studies", "studying", "studied",
    "connect", "connected", "connecting", "connection",
    "create", "creates", "creating", "creation",
    "move", "moving", "moved", "movement",
    "develop", "developed", "developing", "development"
]
```

```python
for word in words:
    print(word+"---->"+lemmatizer.lemmatize(word,pos='v'))
```
