Porter Stemmer

```python
words = [
    "eat", "eating", "eaten", "eats",
    "final", "finally", "finalize", "finalized",
    "play", "playing", "played", "player",
    "study", "studies", "studying", "studied",
    "connect", "connected", "connecting", "connection",
    "create", "creates", "creating", "creation",
    "move", "moving", "moved", "movement",
    "develop", "developed", "developing", "development"
]
```

```python
from nltk import PorterStemmer
stemming=PorterStemmer()

for word in words:
    print(word+"---->"+stemming.stem(word))
```

RegExpStemmer

```python
from nltk.stem import RegexpStemmer
reg_stemmer=RegexpStemmer('ing$|s$|e$|able$')
```

```python
reg_stemmer.stem('eating')
reg_stemmer.stem('ingeating')
```

## Snowball stema

```python
from nltk.stem import SnowballStemmer
snowballStemma=SnowballStemmer('english')
```

```python
for word in words:
    print(word+"----->"+snowballStemma.stem(word))
```
