```python
def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(name="harh",age='32',country='india')
```

```python
### USE args in placxe of any name
def print_names(*Harsh):
    for names in Harsh:
        print(names)
```

```python
print_names(1,2,3,4,5,6)
```

```python
### try,except,finally -> finnaly always run
try:
    num=int(input("Enter the number"))
    result=10/num
except ZeroDivisionError:
    print("enter no greater than 0")
except ValueError:
    print("Enter the correct value")
else:
    print(f"the result is {result}")
finally:
    print("Execution completed")
```
