Tuples in Python

```python
## create a tuple
empty_tuple=()
print(empty_tuple)
print(type(empty_tuple))
```

```python
numbers=tuple([1,2,3,4,5,6])
numbers
## list can ve converted in to tuples and wise-versa is also true
list=list[(1,2,3,4,5,6)]
list
```

```python
mixed_tuples=(1,"Hello-World,3.14")
print(mixed_tuples)
```

```python
## Accesing tuples elements
numbers[0]
numbers[-1]
numbers[0:4]

```

```python
## Tuples Operation
concat_tuples=numbers+mixed_tuples
print(concat_tuples)
 
```

```python
numbers*3
```

```python
## Imutable nature of tuples
## tuples are immutable, which means elements cannot be changed once assigned

```

```python
## tuples method
print(numbers)
print(numbers.count(1))
print(numbers.index(3)) 
```

```python
## Packing and Unpacking of tuples
## packing
packed_tuples=1,"Hello",3,1344
print(packed_tuples)
```

```python
## unpacking of tuples
a,b,c,d=packed_tuples
print(a)
print(b)
print(c)
print(d)
```

```python
## Unpacking with *
numbers=(1,2,3,4,5,6)
first,*middle,last=numbers
print(first)
print(middle)
print(last)


```

```python
## Nested tuples
nested_tuples=((1,2,3),("a","b","c"),(True,False))
## accesing the elememts
print(nested_tuples)
print(nested_tuples[0][2])
print(nested_tuples[1])
```
