```python
empty_dict=dict()
empty_dict
```

```python
student={ ## keys are unique in dict
    "name":"Harsh",
    "age":25,
    "grade":24
}
print(student)
```

```python
## accesing dict elements
student={ ## keys are unique in dict
    "name":"Harsh",
    "age":25,
    "grade":24
}
print(student['grade'])
print(student['age'])


## there is another method also
print(student.get('grade'))




```

```python
## Modifing dict elements
## dict are mutable->you can change the elements in dict
student["age"]=34 
student['address']="India"

del student['grade']
print(student)




```

```python
## Dictionary Methods
keys=student.keys()
print(keys)
values=student.values()
print(values)

items=student.items()
print(items)
```

```python
## shallow copy
student_copy=student
print(student)
print(student_copy)
```
