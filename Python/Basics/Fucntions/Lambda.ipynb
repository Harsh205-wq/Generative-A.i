```python
def even(num):
    if num%2==0:
        return True
    else:
        return False
even(34)
```

```python
even_odd=lambda num:num%2==0
even_odd(67)
```

```python
addition=lambda x,y,z:x+y+z
addition(3,4,5)
```

```python
### map()
numbers = [1,2,3,4,5]

def square(numbers):
    result = []
    for number in numbers:
        result.append(number**2)
    return result

print(square(numbers))
```

```python
list(map(lambda x:x**2,numbers))
```
