While loop

```python
count=0
while count<5:
    print(count)
    count=count+1
```

## loop control statment

```python
## Break statement exists the loop permanently
for i in range(10):
    if i==5:
        break
    print(i)
```

```python
## Continue statment skips the current iteration and continues with the next
for i in range(10):
    if i==5:
        continue
    print(i)
```

```python
## pass statement is a null operation,it does nothing
for i in range(10):
    if i==5:
        pass
    print(i)
```

```python
## nested loop
for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```
