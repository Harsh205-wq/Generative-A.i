```python
from abc import ABC,abstractmethod
class Vechicle(ABC):
    def drive(self):
        print("the vechicle is used for driving")
    @abstractmethod
    def start_engine(self):
        pass
class Car(Vechicle):
    def start_engine(self):
        print("Car engine started")
    
def operate_vechicle(vechicle):
    vechicle.start_engine()
    # vechicle.drive()
    
car=Car()
operate_vechicle(car)
    
    
```
