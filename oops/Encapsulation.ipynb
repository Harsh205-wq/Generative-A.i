```python
class Person:
    def __init__(self,name,age,number):
        self._name=name  ## prptected variable 
        self._age=age    ## protected variable
        self.number=number  ## public variable
        
class Employee(Person):
    def __init__(self, name, age, number):
        super().__init__(name, age, number)

employee=Employee("Harshhh",34,89)
print(employee._name)

# def get_name(person):
#     return person.name
# get_name(person)
```

```python
## getter and setter
class Person:
    def __init__(self,name,age):
        self.__name=name
        self.__age=age
        
     ## getter method   
    def get_name(self):
        return self.__name
    
    ## setter method
    def set_name(self,name):
        self.__name=name
        
person=Person("Krish",33)

print(person.get_name())
print(person.set_name(35))

        
    
```
